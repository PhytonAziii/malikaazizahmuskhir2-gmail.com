## 2.1 Dasar Teori
1. Algoritma

Algoritma merupakan serangkaian langkah yang logis, sistematis, dan terstruktur untuk menyelesaikan suatu permasalahan. Algoritma menjadi dasar dalam proses pemrograman karena membantu menentukan tahapan penyelesaian masalah sebelum diterapkan ke dalam bahasa pemrograman. Algoritma yang baik harus memiliki langkah yang jelas, terurut, dan dapat menghasilkan keluaran sesuai dengan permasalahan yang diberikan.

2. Pemrograman

Pemrograman merupakan proses menerjemahkan algoritma atau solusi suatu permasalahan ke dalam bentuk instruksi yang dapat dipahami dan dijalankan oleh komputer. Melalui pemrograman, suatu algoritma dapat diimplementasikan menjadi sebuah program yang mampu melakukan proses pengolahan data dan menghasilkan keluaran tertentu.

3. Python

Python merupakan bahasa pemrograman tingkat tinggi yang bersifat umum dan memiliki sintaks yang relatif sederhana sehingga mudah dipahami. Python mendukung berbagai konsep pemrograman, seperti variabel, tipe data, operator, percabangan, perulangan, fungsi, serta struktur data. Dalam praktikum algoritma dan pemrograman, Python digunakan sebagai media untuk mengimplementasikan algoritma ke dalam bentuk program.

4. Variabel dan Tipe Data

Variabel merupakan tempat untuk menyimpan data yang digunakan dalam suatu program. Python memiliki berbagai tipe data, di antaranya "integer" untuk bilangan bulat, "float" untuk bilangan desimal, "string" untuk teks, dan "boolean" untuk nilai benar atau salah. Penggunaan variabel dan tipe data yang tepat diperlukan agar data dapat diproses sesuai dengan kebutuhan program.

5. Operator

Operator merupakan simbol yang digunakan untuk melakukan operasi terhadap data atau variabel. Dalam Python terdapat beberapa jenis operator, seperti operator aritmatika, perbandingan, dan logika. Operator aritmatika digunakan untuk melakukan perhitungan, operator perbandingan digunakan untuk membandingkan nilai, sedangkan operator logika digunakan untuk menggabungkan atau memproses kondisi.

6. Percabangan

Percabangan merupakan struktur dalam pemrograman yang digunakan untuk menentukan tindakan berdasarkan kondisi tertentu. Python menyediakan struktur "if", "elif", dan "else" untuk membuat percabangan. Dengan percabangan, program dapat menghasilkan tindakan atau keluaran yang berbeda sesuai dengan kondisi yang terpenuhi.

7. Perulangan

Perulangan merupakan struktur pemrograman yang digunakan untuk menjalankan suatu instruksi secara berulang berdasarkan kondisi atau jumlah pengulangan tertentu. Python menyediakan perulangan "for" dan "while". Penggunaan perulangan dapat membuat program lebih efisien karena instruksi yang sama tidak perlu ditulis berulang kali.

8. Input dan Output

Input merupakan proses memasukkan data ke dalam program, sedangkan output merupakan informasi yang dihasilkan oleh program. Pada Python, fungsi "input()" digunakan untuk menerima masukan dari pengguna, sedangkan fungsi "print()" digunakan untuk menampilkan keluaran ke layar.

9. Pseudocode

Pseudocode merupakan representasi sederhana dari algoritma yang dituliskan menggunakan bahasa yang menyerupai bahasa pemrograman tetapi tidak terikat pada aturan sintaks tertentu. Pseudocode digunakan untuk menggambarkan alur penyelesaian masalah sebelum algoritma diimplementasikan ke dalam bahasa pemrograman Python.

### 2.2.2 Repository GitHub

Setelah program selesai dibuat dan diuji melalui VS Code, selanjutnya dapat diunggah ke GitHub. Repository GitHub berfungsi sebagai tempat penyimpanan kode program secara daring sehingga lebih mudah dikelola dan diakses.

Tahapan yang dilakukan meliputi:

1. Membuat repository baru di GitHub.
2. Membuka folder proyek melalui VS Code.
3. Menjalankan terminal pada VS Code.
4. Menginisialisasi Git pada folder proyek.
5. Menambahkan file proyek ke staging area.
6. Membuat commit untuk menyimpan perubahan.
7. Menghubungkan repository lokal dengan repository GitHub.
8. Melakukan push untuk mengunggah seluruh kode ke GitHub.

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

**Penjelasan :**
1. git init – Menginisialisasi Git pada folder proyek agar menjadi repository lokal.
2. git add . – Menambahkan seluruh file proyek ke staging area.
3. git commit -m "Initial commit" – Menyimpan perubahan awal ke repository dengan pesan Initial commit.
4. git branch -M main – Mengatur nama branch utama menjadi main.
5. git remote add origin ... – Menghubungkan repository lokal dengan repository yang ada di GitHub.
6. git push -u origin main – Mengunggah file dan kode proyek dari repository lokal ke GitHub pada branch main.
