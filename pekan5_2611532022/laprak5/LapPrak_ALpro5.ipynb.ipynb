## 2.1 Dasar Teori
Dalam pemrograman Python, **perulangan** digunakan untuk menjalankan suatu perintah secara berulang sesuai dengan jumlah atau kondisi yang telah ditentukan. Perulangan dapat membantu membuat program menjadi lebih sederhana dan efisien karena programmer tidak perlu menuliskan perintah yang sama secara berulang. Pada praktikum ini, materi yang dipelajari meliputi perulangan `for`, nested loop, serta penggunaan perulangan untuk menentukan dan menghitung jumlah bilangan genap.

**1. Perulangan (Looping)**

Perulangan atau *looping* adalah proses menjalankan suatu perintah secara berulang. Perulangan digunakan ketika suatu proses harus dilakukan beberapa kali sehingga program menjadi lebih singkat dan terstruktur.

**2. Perulangan `for`**

Perulangan `for` digunakan untuk menjalankan suatu perintah berdasarkan urutan atau jumlah pengulangan tertentu. `for` dapat digunakan untuk menampilkan angka secara berurutan, mengolah data, dan melakukan perhitungan.

**3. Nested Loop**

Nested loop adalah perulangan yang berada di dalam perulangan lainnya. Perulangan ini digunakan untuk melakukan proses secara bertingkat, seperti membuat pola angka atau mengolah data yang memiliki baris dan kolom.

**4. Bilangan Genap**

Bilangan genap adalah bilangan yang dapat dibagi dengan 2 tanpa menghasilkan sisa. Contoh bilangan genap yaitu 2, 4, 6, 8, 10, dan seterusnya. Dalam Python, bilangan genap dapat diperiksa menggunakan operator modulus.

**5. Operator Modulus (`%`)**

Operator modulus digunakan untuk mendapatkan sisa dari hasil pembagian. Untuk menentukan bilangan genap, digunakan kondisi `angka % 2 == 0`. Jika hasilnya 0, maka angka tersebut merupakan bilangan genap.

**6. Perhitungan Jumlah Bilangan Genap**

Perulangan dapat digunakan untuk memeriksa angka satu per satu dan menentukan angka yang termasuk bilangan genap. Setiap bilangan genap yang ditemukan akan dihitung menggunakan variabel penghitung. Setelah seluruh angka selesai diperiksa, dapat diketahui jumlah bilangan genap yang ditemukan.

### 2.2.3 Perulangan_for3 "B_2611532022"

```python
# Buat nama file dengan nama perulangan_for3_2611532022.py
# Buat program utuk perulangan for dalam phyton
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_2022
# Program ini menggunakn fungsi input()

ulang_2022 = int(input("Masukkan jumlah perulangan :"))

jumlah_2022 = 0
for i_2022 in range(1, ulang_2022 + 1):
   print(i_2022, end=" ")
   jumlah_2022 = jumlah_2022 + i_2022

if i_2022 < ulang_2022:
  print("+", end=" ")
else:
   print("=", jumlah_2022, end=" ")

print()
print("Jumlah_2022 = ", jumlah_2022)
```

### 2.2.7 nested_for4 "B_2611532022"

```python
# Buat nama file dengan nama nested_for4_2611532022.py
# Buat program utuk perulangan for dalam phyton
# Nama variabel ditambah 4 digit nim terakhir contoh:1234
# Program ini menggunakn fungsi input()

tinggi_2022 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2022 % 2 != 0:
    print("Tinggi harus genap!")
else:
    a_2022 = tinggi_2022
    c_2022 = a_2022
    lebar_2022 = (2 * tinggi_2022) - 2

    for i_2022 in range (1, tinggi_2022 + 1):
        b_2022 = c_2022 + 1

        for j_2022 in range(1, lebar_2022 + 1):

            # Baris atas dan bawah
            if i_2022 == 1 or i_2022 == tinggi_2022:
               if j_2022 == 1 or j_2022 == lebar_2022:
                   print("#", end="")
               else:
                   print("=", end="")

            # Baris isi
            else:
                if j_2022 == 1 or j_2022 == lebar_2022:
                    print("|", end="")
                else:
                    if j_2022 == c_2022:
                        print("<", end="")
                    elif j_2022 == b_2022:
                        print(">", end="")
                    elif j_2022 == (lebar_2022 - c_2022 ):
                        print("<", end="")
                    elif j_2022 == (lebar_2022 - c_2022 + 1):
                        print(">", end="")
                    elif j_2022 > b_2022 and j_2022 < (lebar_2022 - c_2022):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli java
        a_2022 -= 2

        if a_2022 <= 0:
            c_2022 = (-a_2022) + 2
        else:
            c_2022 = a_2022
```

Menampilkan pesan jika tinggi pola yang dimasukkan merupakan bilangan ganjil.

Menyimpan nilai tinggi pola ke dalam variabel a_2022 yang digunakan untuk membantu mengatur bentuk pola.
6. c_2022 = a_2022 →
Digunakan sebagai nilai awal untuk menentukan posisi beberapa karakter pada pola.

Digunakan untuk menentukan lebar pola berdasarkan tinggi yang dimasukkan.

Digunakan untuk memeriksa apakah program sedang membuat baris paling atas atau baris paling bawah.

Menampilkan karakter < pada posisi tersebut.

### 2.2.8 jumlah_genap "B_2611532022"

```python
# Buat nama file dengan nama jumlah_genap_2611532022.py
# Buat program utuk perulangan for dalam phyton
# Nama variabel ditambah 4 digit nim terakhir contoh:1234
# Program ini menggunakn fungsi input()

ulang_2022 = int(input("masukkan nilai batas ;"))

jumlah_genap_2022 = 0
for i_2022 in range(1, ulang_2022 + 1):
    if i_2022 % 2 == 0:
        print(i_2022, end=" ")
        jumlah_genap_2022 = jumlah_genap_2022 + i_2022

        if i_2022 < jumlah_genap_2022:
            print("+", end ="")
        else:
            print("=", jumlah_genap_2022)
print()
print("Jumlah genap = ", jumlah_genap_2022)
```

Digunakan untuk melakukan perulangan dari angka 1 sampai nilai batas yang dimasukkan pengguna.
4. if i_2022 % 2 == 0: →
Digunakan untuk memeriksa apakah angka yang sedang diproses merupakan bilangan genap. Jika hasil i_2022 % 2 adalah 0, maka angka tersebut merupakan bilangan genap.
5. print(i_2022, end=" ") →
Digunakan untuk menampilkan bilangan genap yang ditemukan. end=" " membuat angka tetap berada pada baris yang sama dan dipisahkan dengan spasi.
6. jumlah_genap_2022 = jumlah_genap_2022 + i_2022 →
Digunakan untuk menjumlahkan setiap bilangan genap yang ditemukan. Hasilnya disimpan dalam variabel jumlah_genap_2022.
7. if i_2022 < jumlah_genap_2022: →
Digunakan untuk memeriksa kondisi sebelum menampilkan tanda +.
8. print("+", end="") →
Menampilkan tanda + setelah bilangan genap yang memenuhi kondisi.
9. else: →
Digunakan ketika kondisi pada if tidak terpenuhi.
10. print("=", jumlah_genap_2022) →
Menampilkan tanda = dan hasil penjumlahan bilangan genap yang telah diperoleh.

### 2.2.2 Latihan5 "Latihan_Program_Perulangan_2611532022"

```python
# Program perulangan for untuk membuat segitiga

tinggi_2022 = int(input("Masukkan tinggi segitiga:"))

for i_2022 in range(1, tinggi_2022 + 1):
    for j_2022 in range(1, tinggi_2022 - i_2022 + 1):
        print(" ", end=" ")
    for k_2022 in range(1, 2 * i_2022):
        print("*", end=" ")
    print()
```

**Penjelasan :**
1. tinggi_2022 = int(input("Masukkan tinggi segitiga:")) →
Digunakan untuk meminta pengguna memasukkan tinggi segitiga. int() mengubah input menjadi bilangan bulat.

Merupakan perulangan utama untuk menentukan jumlah baris segitiga. Perulangan dimulai dari 1 sampai nilai tinggi yang dimasukkan.
3. for j_2022 in range(1, tinggi_2022 - i_2022 + 1): →
Merupakan nested loop yang digunakan untuk membuat spasi sebelum tanda bintang. Semakin ke bawah, jumlah spasi semakin sedikit sehingga membentuk segitiga.
4. print(" ", end=" ") →
Digunakan untuk mencetak spasi tanpa langsung berpindah ke baris berikutnya.
5. for k_2022 in range(1, 2 * i_2022): →
Digunakan untuk menentukan jumlah tanda bintang pada setiap baris. Jumlah bintang bertambah pada setiap baris.
6. print("*", end=" ") →
Mencetak tanda * pada baris yang sama dan diberi jarak.
7. print() →
Digunakan untuk berpindah ke baris berikutnya setelah proses pencetakan spasi dan bintang selesai.
