## 2.1 Dasar Teori
Python merupakan salah satu bahasa pemrograman yang banyak digunakan karena memiliki sintaks yang sederhana, mudah dipahami, dan cocok untuk pemula. Dalam membuat sebuah program, diperlukan pemahaman mengenai tipe data, variabel, konstanta, serta cara menerima data dari pengguna. Konsep-konsep tersebut menjadi dasar dalam pembuatan program Python.

1. Boolean

Boolean merupakan tipe data yang hanya memiliki dua nilai, yaitu True dan False. Tipe data ini biasanya digunakan untuk menyatakan suatu kondisi atau hasil dari perbandingan. Contohnya, 10 > 5 akan menghasilkan nilai True, sedangkan 10 < 5 menghasilkan False.

2. String

String adalah tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dapat berupa nama, kalimat, atau simbol dan biasanya ditulis menggunakan tanda kutip tunggal (' ') atau tanda kutip ganda (" "). Contohnya adalah nama = "Andi".

3. Konstanta

Konstanta adalah nilai yang digunakan dan dianggap tetap selama program dijalankan. Python tidak memiliki aturan khusus untuk membuat konstanta, sehingga biasanya konstanta ditulis menggunakan huruf kapital agar mudah dibedakan dari variabel. Contohnya adalah PI = 3.14.

4. Numerik

Numerik merupakan tipe data yang digunakan untuk menyimpan nilai berupa angka. Dalam Python, tipe data numerik yang umum digunakan adalah int untuk bilangan bulat dan float untuk bilangan desimal. Data numerik dapat digunakan dalam berbagai operasi matematika seperti penjumlahan, pengurangan, perkalian, dan pembagian.

5. input

Input merupakan proses memasukkan data dari pengguna ke dalam program. Dalam Python, fungsi input() digunakan untuk menerima masukan dari pengguna. Data yang diperoleh melalui input() secara default berupa string, sehingga perlu dikonversi jika ingin digunakan sebagai angka, misalnya menggunakan int() atau float().

Dengan memahami Boolean, String, Konstanta, Numerik, dan Input, pengguna dapat membuat program Python sederhana yang dapat menerima, menyimpan, dan mengolah berbagai jenis data sesuai kebutuhan.


## 2.2 Kode Program

### 2.2.1 Boolean "B_2611532022"


```python
# Buat file dengan nama Boolean_2611532022.py
#nama variable ditambah 4 digit nim terakhir contoh: nama_1234
# Deklarasi variable dengan tipe data boolean
is_lulus_2022 = True
is_cumlaude_2022 = True

# Menggunakan Boolean
nilai_2022 = 85
batas_lulus_2022 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2022 = nilai_2022 >= batas_lulus_2022 #Hasilnya akan True

print("=== Check Kelulusan === ")
print("Nilai:", nilai_2022)
print("Apakah lulus?:", status_kelulusan_2022)
if is_lulus_2022 and is_cumlaude_2022:
    print("Selamat, Anda lulus dengan prediket cumlaude!")
    
```

**Penjelasan :**
1. Variabel Boolean
is_lulus_2022 dan is_cumlaude_2022 berfungsi menyimpan nilai logika True atau False untuk menunjukkan status kelulusan dan predikat cumlaude. 
2. Variabel Numerik
nilai_2022 berfungsi menyimpan nilai mahasiswa, sedangkan batas_lulus_2022 digunakan sebagai nilai minimum untuk menentukan kelulusan.
3. Menentukan Status Kelulusan
status_kelulusan_2022 = nilai_2022 >= batas_lulus_2022 berfungsi membandingkan nilai mahasiswa dengan batas kelulusan. Hasil perbandingan berupa True atau False.
4. Fungsi print()
Digunakan untuk menampilkan nilai, status kelulusan, dan pesan ke layar.
5. Percabangan if
Berfungsi memeriksa apakah kondisi kelulusan dan cumlaude terpenuhi. Operator and memastikan kedua kondisi harus bernilai True.
6. Operator and
Digunakan untuk menggabungkan dua kondisi. Jika keduanya bernilai True, program akan menampilkan pesan bahwa mahasiswa lulus dengan predikat cumlaude.

### 2.2.2 Input "B_2611532022"

```python
#Buat file dengan nama Input _2611532022.py
#Buat program untuk menghitung luas persegi panjang
#Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
#Program ini menggunakan fungsi input () untuk meminta pengguna
#memasukkan nilai panjang dan lebar peregi panjang.
#Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2022 = input ("masukkan nilai panjang")
lebar_2022 = input ("masukkan nilai lebar")
luas_2022 = int(panjang_2022) * int(lebar_2022)
print("luas =", luas_2022)
```

**Penjelasan :**
1. Fungsi input()
Digunakan untuk meminta pengguna memasukkan nilai panjang dan lebar persegi panjang.
2. Variabel panjang_2022 dan lebar_2022
Berfungsi menyimpan nilai panjang dan lebar yang dimasukkan oleh pengguna.
3. Fungsi int()
Digunakan untuk mengubah data dari input() yang awalnya berupa String menjadi tipe data Integer (bilangan bulat).
4. Perhitungan luas
luas_2022 = int(panjang_2022) * int(lebar_2022) berfungsi menghitung luas persegi panjang dengan rumus panjang × lebar.
5. Fungsi print()
Digunakan untuk menampilkan hasil perhitungan luas persegi panjang ke layar.
6. Hasil program
Program akan meminta pengguna memasukkan panjang dan lebar, kemudian menghitung dan menampilkan luas persegi panjang berdasarkan nilai yang diberikan.

### 2.2.3 Konstanta "B_2611532022"

```python
#Buat file dengan nama Konstanta_2611532022.py
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#Nama variable ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI : Final = 3.14
print("pi: %f" % (PI))
jari_2022 = float(input("Masukkan nilai jari-jari: "))
luas_2022 = PI * jari_2022 * jari_2022
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2022, luas_2022))
```

**Penjelasan :**
1. Import Final
from typing import Final digunakan untuk mendefinisikan nilai yang dianggap sebagai konstanta dan tidak seharusnya diubah.
2. Konstanta PI
PI: Final = 3.14 berfungsi menyimpan nilai π (pi) yang digunakan dalam perhitungan luas lingkaran.
3. Fungsi print()
print("pi: %f" % (PI)) digunakan untuk menampilkan nilai konstanta PI ke layar.
4. Fungsi input() dan float()
jari_2022 = float(input(...)) berfungsi meminta pengguna memasukkan nilai jari-jari. Fungsi float() mengubah input menjadi bilangan desimal.
5. Perhitungan luas lingkaran
luas_2022 = PI * jari_2022 * jari_2022 berfungsi menghitung luas lingkaran menggunakan rumus π × r × r.
6. Menampilkan hasil
print() terakhir berfungsi menampilkan nilai jari-jari dan hasil luas lingkaran. Format %.2f digunakan agar angka ditampilkan dengan dua angka di belakang koma.
7. Hasil program
Program akan menerima nilai jari-jari dari pengguna, menghitung luas lingkaran menggunakan konstanta PI, kemudian menampilkan hasil perhitungannya.

### 2.2.4 Numerik "B_2611532022"

```python
# Buat file dengan nama Numerik_2611532022.py
#nama variable ditambah 4 digit nim terakhir contoh: nama_1234
#create a variable with integer value.
a_2022 = 100
print("The type of variable having value", a_2022, "is", type(a_2022))

#create a variable with float value
b_2022 = 10.2345
print("The type of variable having value", b_2022, "is", type(b_2022))

#create a variable with float value
c_2022 = 100+3j
print("The type of variable having value", c_2022, "is", type(c_2022))

```

**Penjelasan :**
1. Variabel Integer
a_2022 = 100 berfungsi menyimpan bilangan bulat dengan tipe data Integer (int). Fungsi type() digunakan untuk mengetahui tipe data dari variabel tersebut.
2. Variabel Float
b_2022 = 10.2345 berfungsi menyimpan bilangan desimal dengan tipe data Float (float). Fungsi type() digunakan untuk menampilkan tipe data variabel.
3. Variabel Complex
c_2022 = 100+3j berfungsi menyimpan bilangan kompleks dengan tipe data Complex (complex). Huruf j menunjukkan bagian imajiner dari bilangan tersebut.
4. Fungsi print()
Digunakan untuk menampilkan nilai setiap variabel beserta tipe datanya ke layar.
5. Fungsi type()
Berfungsi untuk mengetahui dan menampilkan jenis atau tipe data yang digunakan oleh suatu variabel.
6. Hasil Program
Program menghasilkan informasi bahwa a_2022 bertipe int, b_2022 bertipe float, dan c_2022 bertipe complex.
