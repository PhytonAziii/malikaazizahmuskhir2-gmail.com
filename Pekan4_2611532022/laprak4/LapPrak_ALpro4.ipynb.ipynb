## 2.1 Dasar Teori
Struktur percabangan merupakan bagian dari pemrograman yang digunakan untuk menentukan perintah yang akan dijalankan berdasarkan suatu kondisi. Dengan menggunakan percabangan, program dapat melakukan proses pengambilan keputusan sehingga perintah yang dijalankan dapat berbeda sesuai dengan kondisi yang diberikan.

Dalam Python, terdapat beberapa bentuk struktur percabangan yang dapat digunakan, yaitu if, if-else, if-elif-else, multi-if, dan match-case. Setiap bentuk percabangan memiliki fungsi dan penggunaan yang berbeda sesuai dengan kebutuhan program.

1. Percabangan if

if digunakan untuk menjalankan perintah apabila suatu kondisi bernilai benar (True). Jika kondisi bernilai salah (False), perintah di dalam if tidak akan dijalankan.

2. Percabangan if-else

if-else digunakan untuk kondisi yang memiliki dua kemungkinan. Jika kondisi pada if bernilai benar, perintah pada if akan dijalankan. Jika kondisi salah, program akan menjalankan perintah pada else.

3. Percabangan if-elif-else

if-elif-else digunakan untuk memeriksa beberapa kondisi. Program akan memeriksa kondisi secara berurutan dan menjalankan perintah pada kondisi yang bernilai benar. Jika tidak ada kondisi yang terpenuhi, maka perintah pada else akan dijalankan.

4. Multi-if

Multi-if adalah penggunaan beberapa pernyataan if dalam satu program. Setiap kondisi diperiksa secara terpisah sehingga beberapa kondisi dapat dijalankan apabila masing-masing kondisi bernilai benar.

5. match-case

match-case merupakan struktur percabangan yang digunakan untuk mencocokkan suatu nilai dengan beberapa pola atau pilihan yang telah ditentukan. Struktur ini dapat digunakan ketika program memiliki beberapa pilihan berdasarkan nilai tertentu.

6. Kondisi dan Operator Perbandingan

Kondisi digunakan sebagai dasar dalam pengambilan keputusan pada percabangan. Kondisi dapat menggunakan operator perbandingan seperti >, <, >=, <=, ==, dan !=. Hasil dari perbandingan tersebut berupa True atau False, yang kemudian menentukan perintah yang akan dijalankan oleh program.

### 2.2.1 Program If_elif_else1 "B_2611532022"


```python
# Buat file dengan nama if_elif_else1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakn fungsi input()

umur_2022 = int(input("Input Umur Anda"))
sim_2022 = input("Apakah Anda Sudah Punya SIM C: ")[0]

if umur_2022 >= 17 and sim_2022 == 'y':
    print("Anda Suddah Dewasa dan Boleh Bawa Motor")
elif umur_2022 >= 17 and sim_2022 != 'y':
    print("Anda Suddah Dewasa tetapi Tidak Boleh Bawa Motor")
elif umur_2022 < 17 and sim_2022 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan Tidak Boleh Bawa Motor")
print("Program Selesai")
```

2. **`int()`**

3. **Variabel `umur_2022`**
   Digunakan untuk menyimpan data umur pengguna.

4. **Variabel `sim_2022`**
   Digunakan untuk menyimpan jawaban pengguna mengenai kepemilikan SIM C.

5. **`[0]` (indeks)**
   Digunakan untuk mengambil karakter pertama dari jawaban input SIM C, misalnya `y` dari `yes`.

6. **`if`**
   Digunakan untuk memeriksa kondisi pertama, yaitu apakah umur sudah 17 tahun dan memiliki SIM C.

7. **`elif`**
   Digunakan untuk memeriksa kondisi lain apabila kondisi sebelumnya tidak terpenuhi.

8. **`else`**
   Digunakan untuk menjalankan perintah ketika seluruh kondisi `if` dan `elif` tidak terpenuhi.

9. **Operator `>=`**
   Digunakan untuk membandingkan apakah umur lebih besar atau sama dengan 17.

10. **Operator `<`**
    Digunakan untuk membandingkan apakah umur kurang dari 17.

11. **Operator `==`**
    Digunakan untuk memeriksa apakah nilai `sim_2022` sama dengan `'y'`.

12. **Operator `!=`**
    Digunakan untuk memeriksa apakah nilai `sim_2022` tidak sama dengan `'y'`.

13. **Operator `and`**
    Digunakan untuk menggabungkan dua kondisi yang harus sama-sama benar.

14. **`print()`**
    Digunakan untuk menampilkan hasil sesuai kondisi dan menampilkan pesan **"Program Selesai"** setelah percabangan selesai.

### 2.2.4 Program Match_Case1 "B_2611532022"

```python
# Buat file dengan nama match_case1.print
# Buat program untuk match cas
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakn fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2022 = int(input("Masukkan angka bulan (1-12): "))

match bulan_2022:
    case 1:
        print("Januari")
    case 2:
        print("Feburuari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

**Penjelasan :**
1. **`input()`**
   Digunakan untuk menerima angka bulan yang dimasukkan oleh pengguna.

2. **`int()`**
   Digunakan untuk mengubah input menjadi bilangan bulat agar dapat dicocokkan dengan angka pada setiap `case`.

3. **Variabel `bulan_2022`**
   Digunakan untuk menyimpan angka bulan yang dimasukkan oleh pengguna.

4. **`match`**
   Digunakan untuk mencocokkan nilai `bulan_2022` dengan pilihan yang tersedia.

5. **`case`**
   Digunakan untuk menentukan perintah berdasarkan nilai yang cocok dengan `match`. Contohnya, `case 1` akan menghasilkan **"Januari"**.

6. **`case _`**
   Digunakan sebagai kondisi lainnya atau **default**. Jika angka yang dimasukkan tidak sesuai dengan `case` 1 sampai 12, program akan menampilkan **"Angka Tidak Valid"**.

7. **`print()`**
   Digunakan untuk menampilkan nama bulan berdasarkan angka yang dimasukkan atau menampilkan pesan jika angka tidak valid.

8. **Struktur `match-case`**
   Secara keseluruhan, program menggunakan **`match-case`** untuk mengonversi angka `1–12` menjadi nama bulan. Misalnya, angka `1` menghasilkan **Januari**, angka `2` menghasilkan **Februari**, dan seterusnya.

### 2.2.6 Program Multi_if2 "B_2611532022"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakn fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2022 = float(input("Masukkan total belanja (Rp): "))

# Input satus member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2022 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_2022 = input_member_2022 in ["y", "ya"]
# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2022 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2022 = input_promo_2022 in ["y", "ya"]

total_diskon_persen_2022 = 0

#multi-if terpisah: setiap kondisi diperiksa secara indepeenden
# Diskon bisa ditumpuk (akumulasi) jika memenuhi bebrapa syarat sekaligus

if total_belanja_2022 > 1000000:
    total_diskon_persen_2022 += 10 #Diskon Belanja Besar

if is_member_2022:
    total_diskon_persen_2022 += 5 #Diskon Member

if kode_promo_valid_2022:
    total_diskon_persen_2022 += 15 #Doskon Vocher

#menghitung nominal diskon dan total bayar
nominal_diskon_2022 = total_belanja_2022 * (total_diskon_persen_2022/ 100)
total_bayar_2022 = total_belanja_2022 - nominal_diskon_2022

#output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2022}% (rp {nominal_diskon_2022:,.0f})")
print(f"Total Bayar : Rp {total_bayar_2022:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2022}%")
# output total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. **`input()`**
   Digunakan untuk menerima data dari pengguna, yaitu total belanja, status member, dan status kode promo.

2. **`float()`**
   Digunakan untuk mengubah total belanja menjadi bilangan desimal agar dapat digunakan dalam perhitungan.

3. **Variabel dengan akhiran `_2022`**
   Digunakan sesuai aturan penamaan variabel berdasarkan 4 digit terakhir NIM.

4. **`strip()`**
   Digunakan untuk menghilangkan spasi yang tidak diperlukan pada awal atau akhir input.

5. **`lower()`**
   Digunakan untuk mengubah input menjadi huruf kecil sehingga `Y` atau `y` dapat diproses dengan cara yang sama.

6. **Operator `in`**
   Digunakan untuk memeriksa apakah input pengguna termasuk dalam pilihan `"y"` atau `"ya"`.

7. **Nilai Boolean**
   `is_member_2022` dan `kode_promo_valid_2022` menghasilkan nilai `True` atau `False` berdasarkan jawaban pengguna.

8. **Variabel `total_diskon_persen_2022`**
   Digunakan untuk menyimpan jumlah persentase diskon. Nilai awalnya adalah `0`.

9. **Multi-if**
   Program menggunakan beberapa `if` yang berdiri sendiri. Setiap kondisi diperiksa secara terpisah sehingga diskon dapat **ditumpuk atau diakumulasikan** jika beberapa kondisi terpenuhi.

10. **Operator `>`**
    Digunakan untuk memeriksa apakah total belanja lebih dari `Rp1.000.000`.

11. **Operator `+=`**
    Digunakan untuk menambahkan persentase diskon. Diskon belanja besar `10%`, member `5%`, dan kode promo `15%` dapat dijumlahkan.

12. **Operator `/`**
    Digunakan untuk mengubah persentase diskon menjadi bentuk desimal dalam perhitungan nominal diskon.

13. **Operator `*`**
    Digunakan untuk menghitung nominal diskon berdasarkan total belanja dan persentase diskon.

14. **Operator `-`**
    Digunakan untuk menghitung total pembayaran setelah dikurangi nominal diskon.

15. **F-string**
    Digunakan pada `print(f"...")` untuk menampilkan nilai variabel langsung ke dalam teks.

16. **Format `:,.0f`**
    Digunakan untuk menampilkan angka dengan pemisah ribuan dan tanpa angka desimal, sehingga nominal pembayaran lebih mudah dibaca.

17. **`print()`**
    Digunakan untuk menampilkan rincian pembayaran, total persentase diskon, nominal diskon, dan total yang harus dibayar.

18. **Konsep akumulasi diskon**
    Program dapat memberikan diskon secara bersamaan. Jika pengguna **belanja lebih dari Rp1.000.000, merupakan member, dan memiliki kode promo valid**, maka diskon yang diperoleh adalah **10% + 5% + 15% = 30%**.

19. **Perhitungan total pembayaran**
    Program menghitung nominal diskon terlebih dahulu, kemudian mengurangkannya dari total belanja untuk mendapatkan jumlah yang harus dibayar.


### 2.2.7 Program Nilai "B_2611532022"

```python
# Buat file dengan nama nilai_2611532022.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai = int(input("Inputkan Nilai Angka= "))

if nilai>= 81:
    print("A")
elif nilai >= 70:
    print("B")
elif nilai >= 60:
    print("C")
elif nilai >= 50:
    print("D")
else:
    print("E")
```

   Digunakan untuk menerima nilai angka yang dimasukkan oleh pengguna.

3. **Variabel `nilai`**
   Digunakan untuk menyimpan nilai angka yang dimasukkan oleh pengguna.

4. **`if`**
   Digunakan untuk memeriksa kondisi pertama. Jika nilai lebih besar atau sama dengan `81`, program akan menampilkan **A**.

5. **`elif`**
   Digunakan untuk memeriksa kondisi berikutnya jika kondisi sebelumnya tidak terpenuhi. Pada program ini terdapat beberapa `elif` untuk menentukan rentang nilai.

6. **`else`**
   Digunakan ketika seluruh kondisi `if` dan `elif` tidak terpenuhi. Program akan menampilkan **E**.

7. **Operator `>=`**
   Digunakan untuk membandingkan nilai dengan batas tertentu, seperti `nilai >= 81`, `nilai >= 70`, dan seterusnya.

8. **`print()`**
   Digunakan untuk menampilkan nilai huruf **A, B, C, D, atau E** sesuai dengan nilai angka yang dimasukkan.

9. **Struktur `if-elif-else`**
   Secara keseluruhan, program menggunakan struktur `if-elif-else` untuk mengonversi nilai angka menjadi nilai huruf:

   * **81 ke atas → A**
   * **70–80 → B**
   * **60–69 → C**
   * **50–59 → D**
   * **Di bawah 50 → E**

```python
# Buat file dengan nama Tugas4_2611532022.py
# Program Sistem Loket Alpro Adventure Park
# Nama variabel ditambah 4 digit NIM terakhir

import sys

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# Input Data Pengunjung
nama_2022 = input("Masukkan Nama Pengunjung        : ")
umur_2022 = int(input("Input umur anda                 : "))

sim_2022 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()

jumlah_tiket_2022 = int(input("Masukkan jumlah tiket           : "))

# Validasi jumlah tiket menggunakan if tunggal
if jumlah_tiket_2022 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# Pilihan Paket Wahana
print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba         (Rp 50,000)")
print("2. Arung Jeram          (Rp 75,000)")
print("3. Motor ATV Ekstrim    (Rp 120,000)")
print("4. Roller Coaster Kilat (Rp 100,000)")
print("5. All-Access VIP       (Rp 220,000)")

paket_2022 = int(input("Masukkan nomor paket (1-5) : "))

# Pemilihan wahana menggunakan match-case
match paket_2022:
    case 1:
        nama_wahana_2022 = "Wahana Safari Rimba"
        harga_satuan_2022 = 50000

    case 2:
        nama_wahana_2022 = "Wahana Arung Jeram"
        harga_satuan_2022 = 75000

    case 3:
        nama_wahana_2022 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2022 = 120000

    case 4:
        nama_wahana_2022 = "Wahana Roller Coaster Kilat"
        harga_satuan_2022 = 100000

    case 5:
        nama_wahana_2022 = "Wahana All-Access VIP"
        harga_satuan_2022 = 220000

    case _:
        print("Paket wahana tidak valid!")
        sys.exit()

# Validasi jumlah tiket
if jumlah_tiket_2022 <= 0:
    print("Transaksi dihentikan karena jumlah tiket tidak valid.")
    sys.exit()

# Validasi izin kendali wahana
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2022 == 3:
    if umur_2022 >= 17 and sim_2022 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")

    elif umur_2022 >= 17 and sim_2022 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")

    elif umur_2022 < 17 and sim_2022 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")

    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")

else:
    if umur_2022 >= 10:
        print("Status Akses: Anda memenuhi syarat umur untuk wahana.")
    else:
        print("Status Akses: Anda belum cukup umur untuk wahana ini.")

# Menghitung subtotal
subtotal_2022 = harga_satuan_2022 * jumlah_tiket_2022

# Input Member dan Promo
input_member_2022 = input("Apakah Anda member? (y/t)       : ").strip().lower()
input_promo_2022 = input("Apakah kode promo valid? (y/t) : ").strip().lower()

# Total diskon awal
total_diskon_persen_2022 = 0

# Multi-if untuk diskon akumulasi
if subtotal_2022 >= 200000:
    total_diskon_persen_2022 += 10

if input_member_2022 in ['y', 'ya']:
    total_diskon_persen_2022 += 5

if input_promo_2022 in ['y', 'ya']:
    total_diskon_persen_2022 += 15

if jumlah_tiket_2022 >= 5:
    total_diskon_persen_2022 += 5

# Menghitung nominal diskon dan total bayar
nominal_diskon_2022 = subtotal_2022 * (total_diskon_persen_2022 / 100)
total_bayar_2022 = subtotal_2022 - nominal_diskon_2022

# Audit total pembayaran
if total_bayar_2022 > 300000:
    catatan_layanan_2022 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_2022 = "Terima kasih telah berkunjung."

# Output
print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Nama Pengunjung : {nama_2022}")
print(f"Wahana         : {nama_wahana_2022}")
print(f"Harga Satuan   : Rp {harga_satuan_2022:,.0f}")
print(f"Jumlah Tiket   : {jumlah_tiket_2022}")
print(f"Subtotal Belanja : Rp {subtotal_2022:,.0f}")
print(f"Total Diskon   : {total_diskon_persen_2022}% (Rp {nominal_diskon_2022:,.0f})")
print(f"Total Bayar    : Rp {total_bayar_2022:,.0f}")
print(f"Catatan Layanan: {catatan_layanan_2022}")

print("\nProgram Selesai")
```
